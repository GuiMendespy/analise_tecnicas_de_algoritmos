# Algoritmo do Quadrado Mágico

Dada uma matriz 3x3 deve-se preencher com números de 1 a 9 na finalidade de que a soma das linhas, colunas e das diagonais (Primária e Secundária) devem sempre dar 15.

```python

N = 3 
SOMA = 15  # A soma correta para a matriz 3x3 com números de 1 a 9 é 15

quadrado = [[0] * N for _ in range(N)]
usados = set()


def valido(linha, coluna):
    # Se completou a linha, verifica se a soma dá 15
    if coluna == N - 1:
        if sum(quadrado[linha]) != SOMA:
            return False

    # Se completou a coluna, verifica se a soma dá 15
    if linha == N - 1:
        if sum(quadrado[i][coluna] for i in range(N)) != SOMA:
            return False

    # Quando chega na última célula (2, 2), valida a diagonal principal
    if linha == N - 1 and coluna == N - 1:
        if sum(quadrado[i][i] for i in range(N)) != SOMA:
            return False

    # Quando chega na célula (2, 0), valida a diagonal secundária
    if linha == N - 1 and coluna == 0:
        if sum(quadrado[i][N - 1 - i] for i in range(N)) != SOMA:
            return False

    return True


def backtracking(pos):
    if pos == N * N:
        return True

    linha = pos // N
    coluna = pos % N

    for num in range(1, N * N + 1):
        if num in usados:
            continue

        quadrado[linha][coluna] = num
        usados.add(num)

        if valido(linha, coluna):
            if backtracking(pos + 1):
                return True

        usados.remove(num)
        quadrado[linha][coluna] = 0

    return False


if backtracking(0):
    for linha in quadrado:
        print(*linha)
else:
    print("Nenhuma solução encontrada.")
```
